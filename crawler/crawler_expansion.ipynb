# DidaDex – Quellenexpansion (separat)

Dieses Notebook ergänzt `crawler.ipynb` und ersetzt es **nicht**.
Start: PowerShell im Repository-Hauptordner mit `start_expansion_72h.ps1`.
Hier kann der separate Crawl auch direkt ausgeführt werden.


```python
from pathlib import Path
import subprocess, sys
HERE = Path.cwd().resolve()
ROOT = HERE if (HERE / "engine.py").exists() else HERE.parent
SCRIPT = ROOT / "crawler" / "expansion" / "crawler.py"
assert SCRIPT.exists(), f"Nicht gefunden: {SCRIPT}"
PYTHON = ROOT / ".venv" / "Scripts" / "python.exe"
if not PYTHON.exists(): PYTHON = Path(sys.executable)
print("Repository:", ROOT, "\nPython:", PYTHON)
```

```python
# Vor dem Lauf sicherstellen, dass crawler.ipynb nicht gleichzeitig dieselben Server abruft.
STUNDEN = 72
WORKER = 20
ZIEL_MIB = 500
HOST_DELAY = 2.5
cmd = [str(PYTHON), str(SCRIPT), "run", "--hours", str(STUNDEN),
       "--workers", str(WORKER), "--target-mib", str(ZIEL_MIB),
       "--per-host", "1", "--host-delay", str(HOST_DELAY), "--bpb-sitemap"]
subprocess.run(cmd, cwd=ROOT, check=True)
```

```python
# Status separat ausführen oder nach dem Crawl anzeigen.
subprocess.run([str(PYTHON),str(SCRIPT),"status"],cwd=ROOT,check=True)
```
